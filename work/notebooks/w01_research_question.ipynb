## 2. The question: decision, action, cost of a wrong call

The decision I want to improve is: **which content page should get reviewed and refreshed first?** The person acting on the output is an editor, content strategist, or SEO analyst. They would use the ranking to choose the next pages to inspect, update titles or copy, and decide whether a page deserves a more serious rewrite.

A wrong recommendation has a real cost. A false positive wastes editorial time on pages that do not need attention. A false negative leaves a declining page with demand untouched, which can mean lost clicks and slower recovery. That is why the output should be a ranked queue, not a hard yes/no verdict.

The right language here is decision-support: this work should help someone prioritize effort, not replace judgment.

```python
frame = pd.DataFrame(
    {
        'item': ['decision', 'actor', 'action', 'wrong_call_cost'],
        'value': [
            'which page should be reviewed and refreshed first',
            'editor, content strategist, or SEO analyst',
            'inspect the highest-ranked pages and update titles, copy, or structure',
            'wasted editorial effort on low-opportunity pages or missed recovery on pages with demand',
        ],
    }
)

frame
```

## 4. Careful words: what I can and can't claim

I can claim that my ranking is based on observed signals, that it orders pages in a directional way, and that it is meant to support a decision. I can also claim that the output is testable against future observed outcomes such as later traffic change or later recovery.

I cannot claim causal proof. I cannot claim to "predict Google." I cannot say a page will definitely recover because the score is high. The notebook should stay honest about that boundary: it is a prioritization tool, not a theory of the search engine.

The safest summary is that this project should produce decision-support evidence, not final truth.

```python
claims = pd.DataFrame(
    {
        'claim_type': ['can_claim', 'can_claim', 'cannot_claim', 'cannot_claim'],
        'statement': [
            'This ranking is based on observed signals.',
            'The output is decision-support for prioritizing review.',
            'This notebook does not prove causality.',
            'This notebook does not "predict Google".',
        ],
    }
)

claims
```
